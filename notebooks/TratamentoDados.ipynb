# **2) Carregamento dos dados**

```python
import pandas as pd

url = "https://raw.githubusercontent.com/AnaRaquelCafe/POSTECH_AI_SCIENTIST/main/Base%20de%20dados%20Tech%20Challenge/desafio_nps_fase_1.csv"

df = pd.read_csv(url)

# visualizar estrutura inicial
print(df.shape)
df.head()
```

# **3) Tratamento dos dados**
Primeiro, vamos identificar as colunas numericas vs categóricas e a presença de nulos

```python
df.info()
df.isnull().sum()
```

Agora, podemos executar a limpeza dos valores nulos.
Substituindo valores nulos (NaN) usando:
Moda (mode) → para colunas categóricas (texto)
Mediana (median) → para colunas numéricas

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col].fillna(df[col].mode()[0], inplace=True)
    else:
        df[col].fillna(df[col].median(), inplace=True)
```

Agora podemos então, padronizar o texto para facilitar a analise

```python
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.lower().str.strip()
```

Agora, criamos a variavel da qual vamos analisar (nps_score)

```python
#Função para classificar nps
def classificar_nps(score):
    if score <= 6:
        return "detrator"
    elif score <= 8:
        return "neutro"
    else:
        return "promotor"

df["nps_class"] = df["nps_score"].apply(classificar_nps)
```

Como uma verificação a mais, podemos verificar os duplicados e os excluir

```python
df.duplicated().sum()

#exclui nulos
df = df.drop_duplicates()
```

Salvando e extraindo csv para analise

```python
#salva no colab o arquivo
df.to_csv("dataset_tratado.csv", index=False)

#baixa no computador
from google.colab import files
files.download("dataset_tratado.csv")
```
